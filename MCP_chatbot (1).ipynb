```python

import ollama
from agent_tools import CalculatorTool, web_search_tool

MODEL = "qwen2.5:0.5b"

calculator_tool = CalculatorTool()

tools = [
    {
        "type": "function",
        "function": {
            "name": "calculator",
            "description": calculator_tool.description,
            "parameters": {
                "type": "object",
                "properties": {
                    "expression": {
                        "type": "string",
                        "description": "A mathematical expression to calculate."
                    }
                },
                "required": ["expression"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "web_search",
            "description": "Search the web for current or factual information.",
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {
                        "type": "string",
                        "description": "The web search query."
                    }
                },
                "required": ["query"]
            }
        }
    }
]


def execute_tool(tool_name, arguments):

    if tool_name == "calculator":
        return calculator_tool.forward(arguments["expression"])

    if tool_name == "web_search":
        return web_search_tool.forward(arguments["query"])

    return f"Unknown tool: {tool_name}"


SYSTEM_PROMPT = """
You are a helpful agentic AI assistant.

Use calculator for mathematical calculations.

Use web_search for current information, recent news,
latest events, and facts that may have changed.

Never invent information or tool results.

When you have enough information, provide the final answer.
"""


def run_agent(user_message):

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": user_message
        }
    ]

    for step in range(5):

        print(f"\n--- Agent step {step + 1} ---")

        try:
            response = ollama.chat(
                model=MODEL,
                messages=messages,
                tools=tools
            )

        except Exception as e:
            print("MODEL ERROR:", e)
            return f"MODEL ERROR: {e}"

        message = response["message"]

        print("CONTENT:", message.get("content", ""))
        print("TOOL CALLS:", message.get("tool_calls"))

        tool_calls = message.get("tool_calls")

        if not tool_calls:
            print("\n--- FINAL ---")
            print(message.get("content", ""))
            return message.get("content", "")

        messages.append(message)

        for tool_call in tool_calls:

            tool_name = tool_call["function"]["name"]
            arguments = tool_call["function"]["arguments"]

            print("TOOL:", tool_name)
            print("ARGUMENTS:", arguments)

            result = execute_tool(
                tool_name,
                arguments
            )

            print("RESULT:", result)

            messages.append({
                "role": "tool",
                "tool_name": tool_name,
                "content": str(result)
            })

    return "I couldn't complete the request within the allowed steps."


class AgentWrapper:

    def run(self, prompt):
        return run_agent(prompt)


agent = AgentWrapper()
```
